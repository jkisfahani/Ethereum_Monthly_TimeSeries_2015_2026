## Query 1 — Check block coverage

This query was used to inspect the first and latest block numbers and count the rows in the public Ethereum blocks table.

**Recovered SQL — unchanged:**


```python
SELECT
  MIN(block_number) AS first_block,
  MAX(block_number) AS newest_block,
  COUNT(1) AS total_blocks
FROM `bigquery-public-data.goog_blockchain_ethereum_mainnet_us.blocks`;
```

This query was used to inspect the first and latest transaction timestamps and count transactions.

**Recovered SQL — unchanged:**


```python
SELECT
  MIN(block_timestamp) AS first_transaction,
  MAX(block_timestamp) AS latest_transaction,
  COUNT(*) AS total_transactions
FROM `bigquery-public-data.goog_blockchain_ethereum_mainnet.transactions`;
```
